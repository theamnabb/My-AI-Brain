```python
import numpy as np
import matplotlib.pyplot as plt

sales_data = np.array([
    [1, 200, 300, 250],   # Day 1: [Day, R1, R2, R3]
    [2, 220, 280, 270],   # Day 2
    [3, 210, 320, 260],   # Day 3
    [4, 230, 310, 280],   # Day 4
    [5, 240, 330, 300]    # Day 5
])
```

```python
# Indexing
# Get all sales of Restaurant 2 (column 2).

column_2 = sales_data[:,2]
print("Sales of Restaurant 2:", column_2)
```

```python
# Q2. Slicing

# Get sales data of Day 2 to Day 4 for Restaurant 1 and 3 only.

sliced_data = sales_data[1:4,[1,3]]
print("Sales data from Day 2 to Day 4 for Restaurant 1 and 3:")
print(sliced_data)
```

```python
# Aggregation

# Find the total sales per restaurant.

total_sales = np.sum(sales_data[:, 1:], axis=0)
print("Total Sales per Restaurant:")
print(total_sales)
```

```python
# Minimum & Maximum

# Which restaurant had the highest single-day sales?

# What is the minimum sales of Restaurant 1?

min_sales = np.min(sales_data[:,1], axis=0)
print("=====Minimum Sales per Restaurant=====") 
print("Minimum Sales:", min_sales)


max_sales = np.max(sales_data[:,1:], axis=0)
print("=====Maximum Sales per Restaurant=====") 
print("Maximum Sales:", max_sales)

highest_sale = np.max(sales_data[:,1:])
day, restaurant = np.where(sales_data[:,1:] == highest_sale)
print("Highest single-day sale:", highest_sale)
print("Day:", sales_data[day, 0])
print("Restaurant:", restaurant+1)  # +1 because col 0 = day
```

```python
mask = sales_data[:, 1] > 220
days = sales_data[mask, 0]
print("Days where Restaurant 1 sales > 220:", days)
```

```python
# Reshape

# Reshape the sales_data (excluding Day column) into shape (3, 5) so that each restaurant’s sales are in one row.

reshaped_data = sales_data[:, 1:].reshape(3, 5)
print("Reshaped Sales Data (3 Restaurants, 5 Days):")
print(reshaped_data)

reshaped_data = sales_data[:, 1:].T   # transpose → shape (3, 5)
print("Reshaped Sales Data (3 Restaurants, 5 Days):")
print(reshaped_data)
```

```python
# Data
restaurants = ['R1', 'R2', 'R3']
min_sales = np.min(sales_data[:,1:], axis=0)
max_sales = np.max(sales_data[:,1:], axis=0)

# X positions
x = np.arange(len(restaurants))
# Create a figure
plt.figure(figsize=(10,5))
plt.bar(x - 0.2, min_sales, width=0.4, label='Min Sales', color='skyblue')
plt.bar(x + 0.2, max_sales, width=0.4, label='Max Sales', color='orange')
# Labels and title
plt.title("Minimum and Maximum Sales per Restaurant")
plt.xlabel("Restaurants")
plt.ylabel("Sales")
plt.xticks(x, restaurants)  # set restaurant names on x-axis


# Show the plot
plt.show()
```
