## **Advance Operation with Business examples**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 15000, 16000, 27000, 18000], # Paradise Biryani 
    [2, 20000, 21000, 22000, 23000], # Beijing Bites
    [3, 25000, 26000, 17000, 28000], # Pizza Hub
    [4, 30000, 31000, 32000, 33000], # Burger King
    [5, 35000, 36000, 37000, 38000], # Taco Bell
])
```

```python
# Find the shape

print('shape of sales_data:', sales_data.shape)

# Slicing the data 

print('sales_data:\n',sales_data[1:3])
print('sales_data:\n',sales_data[:,1:])
```

```python
# Sum per year

print("=====Sum per year===== ")
print("2021:",np.sum(sales_data[:, 1:], axis=0))
print("2021:",np.sum(sales_data[:, 1:], axis=1)) # row wise per restaurant
```

```python
# Average 

average_sales = np.mean(sales_data[:, 1:], axis=0)
print("=====Average Sales per Year=====")
print("2021:", average_sales)
```

Minmum sale 

```python
# Minimum sale
min_sales = np.min(sales_data[:,1:], axis=0)
print("=====Minimum Sales per Restaurant=====")
print("Minimum Sales:", min_sales)

# MAximum sale
max_sales = np.max(sales_data[:,1:], axis=0)    
print("=====Maximum Sales per Restaurant=====")
print("Maximum Sales:", max_sales)
```
