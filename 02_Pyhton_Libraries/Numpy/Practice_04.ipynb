```python
import numpy as np

sales_data = np.array([
    [1, 200, 300, 250],   # Day 1: [Day, R1, R2, R3]
    [2, 220, 280, 270],   # Day 2
    [3, 210, 320, 260],   # Day 3
    [4, 230, 310, 280],   # Day 4
    [5, 240, 330, 300]    # Day 5
])
```

```python
# Indexing
# Get all sales of Restaurant 2 (column 2).

column_2 = sales_data[:,2]
print("Sales of Restaurant 2:", column_2)
```

```python
# Q2. Slicing

# Get sales data of Day 2 to Day 4 for Restaurant 1 and 3 only.

sliced_data = sales_data[1:4,[1,3]]
print("Sales data from Day 2 to Day 4 for Restaurant 1 and 3:")
print(sliced_data)
```

```python
# Aggregation

# Find the total sales per restaurant.

total_sales = np.sum(sales_data[:, 1:], axis=0)
print("Total Sales per Restaurant:")
print(total_sales)
```

```python
# Minimum & Maximum

# Which restaurant had the highest single-day sales?

# What is the minimum sales of Restaurant 1?

min_sales = np.min(sales_data[:,1], axis=0)
print("=====Minimum Sales per Restaurant=====") 
print("Minimum Sales:", min_sales)


max_sales = np.max(sales_data[:,1:], axis=0)
print("=====Maximum Sales per Restaurant=====") 
print("Maximum Sales:", max_sales)

highest_sale = np.max(sales_data[:,1:])
day, restaurant = np.where(sales_data[:,1:] == highest_sale)
print("Highest single-day sale:", highest_sale)
print("Day:", sales_data[day, 0])
print("Restaurant:", restaurant+1)  # +1 because col 0 = day
```
