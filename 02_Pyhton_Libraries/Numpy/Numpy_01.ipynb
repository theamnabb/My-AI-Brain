### **import numpy**

```python
import numpy as np
```

### **Creating Array**

```python
array_01 = np.array([1,2,3,4,5])
print('Array one: ',array_01) 

array_02 = np.array([[1,2,3,4],[5,6,7,8]])
print('Array Two: ', array_02)
```

### **Python List vs Numpy Array**

```python
pyhton_list = [1,2,3,4,5]
print('Python List: ', pyhton_list * 2) 

numpy_array = np.array([1,2,3,4,5])
print('Numpy Array: ', numpy_array * 2)
```

### **Creating Array from Scratch**

```python
zero = np.zeros((2,3))
print('Zero Array:\n ', zero, '\n') #Zero Array 

ones = np.ones((4,5))
print('ones Array:\n ', ones, '\n') #ones Array  | row by column

full = np.full((2,3), 7)
print('Full Array:\n ', full, '\n') #Full Array

random = np.random.random((3,4))
print('Random Array:\n ', random, '\n') #Random Array


sequence = np.arange(10, 40, 2)
print('Sequence Array:\n ', sequence, '\n') #Sequence Array


random_integer = np.random.randint(low = 0 , high = 40 , size = (3 , 5))
print("Random_integer List are \n" , random_integer , "\n" )

```

### **Vector , Matrix , Tensor**

```python
vector = np.array([1,2,3,4,5])
print('Vector:\n ', vector, '\n') #Vector

matrix = np.array([[1,2,3],[4,5,6]])
print('Matrix:\n ', matrix, '\n') #Matrix

tensor = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print('Tensor:\n ', tensor, '\n') #Tensor
```

## **Properties of Array**

```python
array = np.array([[1,2,3,4],[5,6,7,8]])
array2 = np.array([[1,2,3,4],[5,6,7,8.6]])

print('Shapes of Array:', array.shape) #Shapes of Array
print('Data Type of Array:', array.dtype) #Data Type of Array
print('Data Type of Array:', array2.dtype) #Data Type of Array
print('Size of Array:', array.size) #Size of Array
print('Number of Dimensions:', array.ndim) #Number of Dimensions
print('Item Size:', array.itemsize) #Item Size
```
