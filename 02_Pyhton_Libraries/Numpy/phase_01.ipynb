### **import numpy**

```python
import numpy as np
import time
```

### **Creating Array**

```python
array_01 = np.array([1,2,3,4,5])
print('Array one: ',array_01) 

array_02 = np.array([[1,2,3,4],[5,6,7,8]])
print('Array Two: ', array_02)
```

### **Python List vs Numpy Array**

```python
pyhton_list = [1,2,3,4,5]
print('Python List: ', pyhton_list * 2) 

numpy_array = np.array([1,2,3,4,5])
print('Numpy Array: ', numpy_array * 2)
```

### **Creating Array from Scratch**

```python
zero = np.zeros((2,3))
print('Zero Array:\n ', zero) #Zero Array 

ones = np.ones((4,5))
print('ones Array:\n ', ones) #ones Array  | row by column

full = np.full((2,3), 7)
print('Full Array:\n ', full) #Full Array

random = np.random.random((3,4))
print('Random Array:\n ', random) #Random Array


sequence = np.arange(10, 40, 2)
print('Sequence Array:\n ', sequence) #Sequence Array


random_integer = np.random.randint(low = 0 , high = 40 , size = (3 , 5))
print("Random_integer List are \n" , random_integer , "\n" )
```
