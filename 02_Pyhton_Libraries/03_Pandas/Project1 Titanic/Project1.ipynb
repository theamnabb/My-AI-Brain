Load the CSV file into a Pandas DataFrame.

Print the first 5 rows to see the data.

Print the shape of the dataset (number of rows and columns).

```python
import pandas as pd

# To load a file 

df = pd.read_csv('D:/My-AI-Brain/02_Pyhton_Libraries/03_Pandas/Project1 Titanic/train.csv',encoding='Latin1')


rows= df.head()
print(rows)

print("Dataset shape:", df.shape)

print('Missing Value', df.isnull().sum())
```

```python
print(df['Survived'].value_counts())
```

```python
# Group by 'Sex' and calculate mean of 'Survived'
survival_by_gender = df.groupby('Sex')['Survived'].mean()
print(survival_by_gender)
```

```python
# Group by 'Pclass' and calculate mean of 'Survived'
survival_by_class = df.groupby('Pclass')['Survived'].mean()
print(survival_by_class)
```

```python
print(df['Age'].isnull().sum())
```

```python
# Fill missing Age with mean
print(df['Age'].fillna(df['Age'].mean(), inplace=True))
print(df['Age'].isnull().sum())
print(df.tail())
```

```python
# Drop Cabin column
df.drop(columns=['Cabin'], inplace=True)

# Check the columns to confirm
print(df.columns)
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
print(df[['SibSp', 'Parch', 'FamilySize']].head())
```

```python
import matplotlib.pyplot as plt

# Survival rate by gender
df.groupby('Sex')['Survived'].mean().plot(kind='bar')
plt.title("Survival Rate by Gender")
plt.ylabel("Survival Rate")
plt.show()

# Age distribution
df['Age'].hist(bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```
