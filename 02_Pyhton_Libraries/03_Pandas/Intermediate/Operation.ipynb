# **Modify the Data**

```python
# Create a dataFRame from dictionary
import pandas as pd
data = {
    'Name':['Tom', 'Nick', 'Krish', 'Jack'],
    'Age':[20, 21, 19, 18],
    "City":['New York', 'Los Angeles', 'Chicago', 'Houston'],
    "Gender":['M', 'M', 'M', 'M'],
    "occupation":['Student', 'Engineer', 'Doctor', 'Artist']
}
df =pd.DataFrame(data)
print(df)
```

# **Add new Column**

```python
df['Status'] = ['Single', 'Married', 'Single', 'Married']
print(df)
```

```python
# By using insert function 

df.insert(2, 'Country', ['US', 'USA', 'UK', 'Dubai'])
print(df)
```

# **Drop the column** 

```python
df.drop('Country', axis=1, inplace=True)
print(df)
```

# **Adding Rows**

```python
df.loc[len(df)] = ['David', 40, 'Houston','M', 'Lawyer', 'Married']
print(df)
```

# **Handle Missing data in dataset**

```python
# isnull() for missing value

import pandas as pd

data = {
    'Name': ['Alice', 'Bob', None, 'David'],
    'Age': [25, 70, 30, 40],
    'City': ['New York', None, 'Chicago', 'Houston']
}
df = pd.DataFrame(data)

print(df)

print("\nCheck missing values:\n", df.isnull())

```

```python
# Count missing values per column:
df.isnull().sum()
```

## **dropna()**

```python
df_drop_rows = df.dropna()
print("\nAfter dropping rows with NaN:\n", df_drop_rows)
```

```python
df_drop_colum = df.dropna(axis=1)
print("\nAfter dropping columns with NaN:\n", df_drop_colum)
```

```python
df_fill = df.fillna("Unknown") # here we can also use calculate like  mean median etc
print("\nFill NaN with 'Unknown':\n", df_fill)
```

# **Sorting**

```python
df_sorting = df.sort_values(by='Age')
print("\nDataFrame sorted by Age:\n", df_sorting)
```

```python
df.sort_values(by='Age', ascending=False) # Descending order    
```

```python

print(df['Age'].mean())
print(df['Age'].median())
print(df['Age'].sum())
print(df['Age'].min())
print(df['Age'].max())  
print(df['Age'].std())  # Standard Deviation
print(df['Age'].count()) # Count of non-null values
```
