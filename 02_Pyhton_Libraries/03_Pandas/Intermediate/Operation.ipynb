# **Modify the Data**

```python
# Create a dataFRame from dictionary
import pandas as pd
data = {
    'Name':['Tom', 'Nick', 'Krish', 'Jack'],
    'Age':[20, 21, 19, 18],
    "City":['New York', 'Los Angeles', 'Chicago', 'Houston'],
    "Gender":['M', 'M', 'M', 'M'],
    "occupation":['Student', 'Engineer', 'Doctor', 'Artist']
}
df =pd.DataFrame(data)
print(df)
```

# **Add new Column**

```python
df['Status'] = ['Single', 'Married', 'Single', 'Married']
print(df)
```

```python
# By using insert function 

df.insert(2, 'Country', ['US', 'USA', 'UK', 'Dubai'])
print(df)
```
