# **Add a new Column**

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
df['City'] = ['New York', 'Los Angeles', 'Chicago',]
print(df)

# Add column on condition
df['Salary'] = df['Age'] * 1000
print(df)
df.insert(1, 'Gender', ['F', 'M', 'M'])
```

# **Dropping/ Removing column**

```python
df.drop('Salary', axis=1, inplace=True, errors='ignore')
print(df)

# Multiple columns
df.drop(['Age', 'City'], axis=1, inplace=True, errors='ignore')
print(df)

#add new column 

df.insert(2, 'Country', ['USA', 'UK', 'USA'])
print(df)
```

# **Dropping/ Removing row**

```python
df = df.insert(0, 'Country', ['USA', 'UK', 'USA'])
print(df)

df['Name']=['Alice', 'Bob', 'Charlie']
print(df)
```

# **Rename the Column**

```python
df = df.rename(columns={'Name': 'Full_Name'})
print(df)
```
