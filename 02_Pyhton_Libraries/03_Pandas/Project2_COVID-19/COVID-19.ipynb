# **COVID-19**

```python
import pandas as pd

# Load file

df = pd.read_csv('D:/My-AI-Brain/02_Pyhton_Libraries/03_Pandas/Project2_COVID-19/country_wise_latest.csv',encoding='Latin1')

print(df.head())

print(df.shape)

print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
print(df['Deaths'].describe())
```

```python
country_totals = df.groupby('Country/Region')[['Confirmed', 'Deaths']].sum()

top_countries = country_totals.sort_values(by='Confirmed', ascending=False)

top_5 = top_countries.head(5)

print(top_5)
```

```python
import matplotlib.pyplot as plt

# Top 5 countries 

# Bar chart for Confirmed cases
top_5['Confirmed'].plot(kind='bar', color='skyblue')
plt.title("Top 5 Countries by Confirmed COVID-19 Cases")
plt.ylabel("Confirmed Cases")
plt.xlabel("Country")
plt.show()

# Bar chart for Deaths
top_5['Deaths'].plot(kind='bar', color='salmon')
plt.title("Top 5 Countries by COVID-19 Deaths")
plt.ylabel("Deaths")
plt.xlabel("Country")
plt.show()
```
